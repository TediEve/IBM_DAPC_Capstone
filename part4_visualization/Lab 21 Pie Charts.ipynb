##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
# Print count to verify data exists before plotting
df_db_exploded = (
    df["DatabaseWantToWorkWith"]
    .astype(str)
    .replace("nan", pd.NA)
    .dropna()
    .str.split(";")
    .explode()
    .str.strip()
)

top_5_db = df_db_exploded.value_counts().head(5)

# Render Pie Chart
plt.figure(figsize=(8, 8))

plt.pie(
    top_5_db.values,
    labels=top_5_db.index,
    autopct="%1.1f%%",
    startangle=140,
    colors=sns.color_palette("Set2", n_colors=len(top_5_db)),
    wedgeprops={"edgecolor": "black", "linewidth": 0.5},
)

plt.title(
    "Top 5 Most-Desired Databases",
    fontsize=13,
    fontweight="bold",
    pad=20,
)

plt.tight_layout()
plt.show()
```

The most-desired database by far is PostgreSQL with 33.5 % almost double the second popular SQLite with its 18.8%.

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.



```python
##Write your code here
# Extract top 5 developer roles
top_5_devs = df["DevType"].dropna().value_counts().head(5)

# Render Pie Chart using Set2 palette
plt.figure(figsize=(8, 8))

plt.pie(
    top_5_devs.values,
    labels=top_5_devs.index,
    autopct="%1.1f%%",
    startangle=140,
    colors=sns.color_palette("Set2", n_colors=len(top_5_devs)),
    wedgeprops={"edgecolor": "black", "linewidth": 0.5},
)

plt.title(
    "Top 5 Developer Roles (DevType)",
    fontsize=13,
    fontweight="bold",
    pad=20,
)

plt.tight_layout()
plt.show()
```

With 46.7 % Developer, full-stack is the clear winner in the most popular developer roles almost 7.5 times more represented than the fifth developer, desktop and enterprise apps with 6.4%

##### 1.3 Create a pie chart for the operating systems used by respondents for professional use


The `OpSysProfessional` use column shows the operating systems developers use professionally. Let’s visualize the distribution of the top operating systems in a pie chart.



```python
##Write your code here
# Split semi-colon separated operating systems and explode into individual rows
df_opsys_exploded = (
    df["OpSysProfessional use"]
    .astype(str)
    .replace("nan", pd.NA)
    .dropna()
    .str.split(";")
    .explode()
    .str.strip()
)

# Extract top operating systems
top_opsys = df_opsys_exploded.value_counts().head(5)

# Render Pie Chart
plt.figure(figsize=(8, 8))

plt.pie(
    top_opsys.values,
    labels=top_opsys.index,
    autopct="%1.1f%%",
    startangle=140,
    colors=sns.color_palette("Set2", n_colors=len(top_opsys)),
    wedgeprops={"edgecolor": "black", "linewidth": 0.5},
)

plt.title(
    "Top Operating Systems Used Professionally",
    fontsize=13,
    fontweight="bold",
    pad=20,
)

plt.tight_layout()
plt.show()
```

**Windows dominance:** Remains the top corporate standard worldwide, largely due to deep integration with enterprise infrastructure, active directory setups, and broad hardware accessibility.

**macOS adoption:** Highly concentrated among developers in North America and Western Europe, heavily driven by mobile (iOS) development requirements, Unix-based terminal features, and company hardware provisioning.

**Linux ecosystem breadth:** Combined across distributions like Ubuntu, Debian, and RHEL, Linux represents a huge chunk of the development landscape (~40%+). Its strength is driven by server-side development, containerization (Docker, Kubernetes), and open-source tooling.
