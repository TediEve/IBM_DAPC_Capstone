### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>
The functions below will download the dataset into your browser:
</p>


```python
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
raw_full_dups = df.duplicated().sum()
raw_id_dups = df.duplicated(subset=["ResponseId"]).sum()

print("--- Raw Dataset (df) ---")
print(f"Total Rows: {len(df):,}")
print(f"Full-Row Duplicates: {raw_full_dups}")
print(f"Duplicate ResponseIds: {raw_id_dups}")
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
## Write your code here
df.drop_duplicates()
clean_full_dups = df.duplicated().sum()
clean_id_dups = df.duplicated(subset=["ResponseId"]).sum()

print("--- Raw Dataset (df) ---")
print(f"Total Rows: {len(df):,}")
print(f"Full-Row Duplicates: {clean_full_dups}")
print(f"Duplicate ResponseIds: {clean_id_dups}")
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
## Write your code here
df.isna().sum()
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
## Write your code here
df['RemoteWork'].isna().sum()
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
## Write your code here
df['RemoteWork'].value_counts()
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
## Write your code here
mode_remote = df['RemoteWork'].mode()[0]
print(f'Most frequent work mode: {mode_remote}')
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
## Write your code here
df['RemoteWork'] = df['RemoteWork'].fillna(mode_remote)
#verify
print('Missing values after inputing', df['RemoteWork'].isna().sum())
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
# Identify compensation-related columns
comp_cols = [
    col
    for col in df.columns
    if any(keyword in col.lower() for keyword in ["comp", "salary", "pay"])
]

print(f"Identified Compensation Columns ({len(comp_cols)}):")
print(comp_cols)
print("-" * 65)

# Display missing counts and statistical summary
for col in comp_cols:
    missing_cnt = df[col].isnull().sum()
    missing_pct = (missing_cnt / len(df)) * 100
    print(
        f"Column: {col} | Missing: {missing_cnt:,} rows ({missing_pct:.2f}%)"
    )

print("\n--- Numerical Summary Statistics ---")
display(df[comp_cols].describe(include="all").T)
```
