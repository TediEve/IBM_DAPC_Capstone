### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
df['Employment'].value_counts()
```

As this column is a multiple choice category we will do some preprocessing before using One-hot encoding.

```python
## Write your code here
target_col = "Employment"

# Generate one-hot encoded binary columns
employment_encoded = df[target_col].str.get_dummies(sep=";")

# Clean column names (add prefix to avoid column name collisions)
employment_encoded.columns = [
    f"Employment_{col.strip().replace(' ', '_').replace(',', '')}"
    for col in employment_encoded.columns
]

# Concatenate the new binary indicator columns back to df
df = pd.concat([df, employment_encoded], axis=1)

print("\nFirst 5 rows preview of encoded features:")
display(df[employment_encoded.columns].head())
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


Checking skewness is the standard way to decide between mean and median:
1. Near 0 Skew (Symmetric): Use Mean (e.g., standard continuous metrics).
2. High Skew ($> \vert{}1.0\vert{}$): Use Median (e.g., compensation or highly skewed counts with extreme outliers).

```python
import numpy as np
# get all numeric columns
numeric_cols = df.select_dtypes(include=[np.number]).columns

for col in numeric_cols:
    missing_cnt = df[col].isnull().sum()
    if missing_cnt > 0:
        # Check skewness: High skew (|skew| > 1.0) -> Median, Symmetric -> Mean
        if abs(df[col].skew()) > 1.0:
            fill_val = df[col].median()
            strategy = "Median"
        else:
            fill_val = df[col].mean()
            strategy = "Mean"

        df[col] = df[col].fillna(fill_val)
        print(f"Imputed '{col}' ({missing_cnt:,} missing) using {strategy}: {fill_val:,.2f}")

print("\n--- Imputation Complete ---")
print(f"Remaining missing numeric values: {df[numeric_cols].isnull().sum().sum()}")
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
# get categorical columns (object and string/category dtypes)
categorical_cols = df.select_dtypes(
    include=["object", "string", "category"]
).columns

# loop through and impute missing values with the mode
for col in categorical_cols:
    missing_cnt = df[col].isnull().sum()
    if missing_cnt > 0:
        col_mode = df[col].mode()[0]
        df[col] = df[col].fillna(col_mode)

print("\n--- Categorical Imputation Complete ---")
print(
    f"Remaining missing categorical values: {df[categorical_cols].isnull().sum().sum()}"
)
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


Min-Max scaling rescales data into a strict $[0, 1]$ interval using the formula:$$\text{Scaled Value} = \frac{X - X_{\min}}{X_{\max} - X_{\min}}$$

```python
comp_min = df["ConvertedCompYearly"].min()
comp_max = df["ConvertedCompYearly"].max()

df["ConvertedCompYearly_MinMax"] = (df["ConvertedCompYearly"] - comp_min) / (
    comp_max - comp_min
)
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
# Apply natural log transformation: ln(1 + ConvertedCompYearly)
df['ConvertedCompYearly_Log'] = np.log1p(df['ConvertedCompYearly'])

# Evaluate skewness reduction
raw_skew = df['ConvertedCompYearly'].skew()
log_skew = df['ConvertedCompYearly_Log'].skew()

print("--- 6.2 Log Transformation Summary ---")
print(f"Original Skewness:   {raw_skew:.2f}")
print(f"Log-Transformed Skew:{log_skew:.2f}")
```

```python
# Base-10 Log Transformation: log10(1 + x)
df['ConvertedCompYearly_Log10'] = np.log10(df['ConvertedCompYearly'] + 1)
```

```python
log10_skew = df['ConvertedCompYearly_Log10'].skew()
print(f'Log10-Transformed Skew: {log10_skew}')
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
df['YearsCodePro'].value_counts()
```

```python
# clean YearsCodePro (convert survey text inputs to float)
years_map = {
    "Less than 1 year": 0.5,
    "More than 50 years": 51.0,
}

df["YearsCodePro_Clean"] = (
    df["YearsCodePro"]
    .replace(years_map)
    .pipe(pd.to_numeric, errors="coerce")
)

# If any missing values exist in YearsCodePro_Clean, fill with median
df["YearsCodePro_Clean"] = df["YearsCodePro_Clean"].fillna(
    df["YearsCodePro_Clean"].median()
)

# define experience bin boundaries and labels
# Tiers: Entry (0-2 yrs), Junior (2-5 yrs), Mid (5-10 yrs), Senior (10-20 yrs), Executive/Lead (20+ yrs)
bins = [-np.inf, 2, 5, 10, 20, np.inf]
labels = ["Entry-level", "Junior", "Mid-level", "Senior", "Executive / Lead"]

# Create ExperienceLevel column using pd.cut
df["ExperienceLevel"] = pd.cut(
    df["YearsCodePro_Clean"],
    bins=bins,
    labels=labels,
    right=True,
)

display(df["ExperienceLevel"].value_counts(dropna=False).sort_index())

display(df[["YearsCodePro", "YearsCodePro_Clean", "ExperienceLevel"]].head())
```
