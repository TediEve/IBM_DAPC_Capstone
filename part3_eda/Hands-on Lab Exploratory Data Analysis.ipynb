#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
## Write your code here
# check for missing values in critical columns
critical_cols = ['Employment', 'JobSat', 'RemoteWork']
print('missing values count')
display(df[critical_cols].isna().sum())
# check and missing percentages
print('Missing values percentage')
display(df[critical_cols].isna().sum()*100/len(df))
```

```python
# ccheck values in JobSat and RemoteWork
print("values JobSat")
display(df['JobSat'].value_counts())
print("values RemoteWork")
display(df['RemoteWork'].value_counts())
```

As there are 55.5% missing values in the JobSat column imputing with mean or median will distort the distribution. We can impute with -1 to keep the column numeric or just drop the rows dynamicaly when doing analysis using this column, as droping permannently 55% of our sample will destroy valuable data.
With only 16.24 % missing in RemoteWork imputing with the mode is a sound choice.

```python
# imputing RemoteWork with the most frequent
df['RemoteWork'] = df['RemoteWork'].fillna(df['RemoteWork'].mode()[0])
# check missing values
print(f"Missing values after imputing {df['RemoteWork'].isna().sum()}")
```

```python
# check values in RemoteWork after imputing
df['RemoteWork'].value_counts()
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
# drop rows with missing values in JobSat
jobsat_df = df.dropna(subset=['JobSat'])
# check value consistency in YearsCodePro
jobsat_df['YearsCodePro'].value_counts()
```

We have some non numeric values which we should map beforehand.

```python
# map non-numeric values in YearsCodePro
years_map = {
    "Less than 1 year": 0.5,
    "More than 50 years": 51.0,
}

jobsat_df["YearsCodePro_Numeric"] = (
    jobsat_df["YearsCodePro"]
    .replace(years_map)
    .pipe(pd.to_numeric, errors="coerce")
)
# create experience ranges for YearsCodePro_Numeric
bins = [0, 5, 10, 20, 100]
labels = ["0-5 years", "5-10 years", "10-20 years", "20+ years"]
jobsat_df["ExperienceRange"] = pd.cut(
    jobsat_df["YearsCodePro_Numeric"],
    bins=bins,
    labels=labels,
    right=True,
)
```

```python
# Calculate the median JobSat for each range.
satisfaction_by_exp = (
    jobsat_df.groupby("ExperienceRange")["JobSat"]
    .agg( "median")
    .reset_index()
)
satisfaction_by_exp
```

```python
# Set visual style
sns.set_theme(style="whitegrid", font="sans-serif")

# Plot
plt.figure(figsize=(10, 5))
ax = sns.barplot(
    data=satisfaction_by_exp,
    x="ExperienceRange",
    y="JobSat",
    palette="Blues_d",
    hue="ExperienceRange",
    legend=False,
)

# Customize plot titles and labels
plt.title(
    "Median Job Satisfaction across Career Experience Tiers",
    fontsize=12,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Years of Professional Coding Experience", fontsize=10)
plt.ylabel("Mean Job Satisfaction (0-10 Scale)", fontsize=10)
plt.ylim(0, 10)  # Keep full 0-10 satisfaction scale

# Annotate exact score values above each bar
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(
            f"{height:.2f}",
            (p.get_x() + p.get_width() / 2.0, height),
            ha="center",
            va="bottom",
            fontsize=10,
            fontweight="bold",
            xytext=(0, 3),
            textcoords="offset points",
        )

plt.tight_layout()
plt.show()
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
# Set visual theme
sns.set_theme(style="whitegrid", font="sans-serif")

# Ensure JobSat is sorted numerically (0.0 to 10.0)
jobsat_df["JobSat_Sorted"] = jobsat_df["JobSat"].astype(int)

plt.figure(figsize=(12, 5))

# Render Count Plot
ax = sns.countplot(
    data=jobsat_df,
    x="JobSat_Sorted",
    palette="viridis",
    hue="JobSat_Sorted",
    legend=False,
)

# Titles and Axis Labels
plt.title(
    "Distribution of Developer Job Satisfaction (0-10 Rating Scale)",
    fontsize=13,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Job Satisfaction Rating (0 = Extremely Dissatisfied, 10 = Extremely Satisfied)", fontsize=10)
plt.ylabel("Number of Respondents", fontsize=10)

# Format Y-axis with comma thousand separators
ax.yaxis.set_major_formatter("{x:,.0f}")

# Annotate exact count and percentage above each bar
total_responses = len(jobsat_df)

for p in ax.patches:
    height = p.get_height()
    if height > 0:
        pct = (height / total_responses) * 100
        ax.annotate(
            f"{pct:.1f}%",
            (p.get_x() + p.get_width() / 2.0, height),
            ha="center",
            va="bottom",
            fontsize=9,
            xytext=(0, 3),
            textcoords="offset points",
        )

# Expand Y-axis slightly to fit top text annotations comfortably
plt.ylim(0, jobsat_df["JobSat_Sorted"].value_counts().max() * 1.15)
plt.tight_layout()
plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
# Set visual theme
sns.set_theme(style="whitegrid", font="sans-serif")

plt.figure(figsize=(12, 5))

# Render Count Plot
ax = sns.countplot(
    data=jobsat_df,
    x="RemoteWork",
    palette="viridis",
    hue="RemoteWork",
    legend=False,
)

# Titles and Axis Labels
plt.title(
    "Distribution of Remote Work Preference",
    fontsize=13,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Remote Work Option", fontsize=10)
plt.ylabel("Number of Respondents", fontsize=10)

# Format Y-axis with comma thousand separators
ax.yaxis.set_major_formatter("{x:,.0f}")

# Annotate exact count and percentage above each bar
total_responses = len(jobsat_df)

for p in ax.patches:
    height = p.get_height()
    if height > 0:
        pct = (height / total_responses) * 100
        ax.annotate(
            f"{pct:.1f}%",
            (p.get_x() + p.get_width() / 2.0, height),
            ha="center",
            va="bottom",
            fontsize=9,
            xytext=(0, 3),
            textcoords="offset points",
        )

# Expand Y-axis slightly to fit top text annotations comfortably
plt.ylim(0, jobsat_df['RemoteWork'].value_counts().max() * 1.15)
plt.tight_layout()
plt.show()
```

```python
# Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.
# we use the whole dataset df as JobSat is not in use now
crosstab_emp = pd.crosstab(
    df['Employment'], 
    df['RemoteWork'], 
    normalize='index'  
) * 100

print("--- Remote Work Preference by Employment Status (%) ---")
display(crosstab_emp.round(2))

#  Cross-tabulation: RemoteWork vs. Job Role (DevType)
if 'DevType' in df.columns:
    crosstab_role = pd.crosstab(
        df['DevType'], 
        df['RemoteWork'], 
        normalize='index'
    ) * 100

    print("\n--- Remote Work Preference by Top Job Roles (%) ---")
    # Display top 10 most common roles for readability
    top_roles = df['DevType'].value_counts().head(10).index
    display(crosstab_role.loc[top_roles].round(2))
```

```python
sns.set_theme(style="whitegrid", font="sans-serif")

# Filter for top job roles to keep the visualization clear
if 'DevType' in df.columns:
    top_roles = df['DevType'].value_counts().head(8).index
    role_df = df[df['DevType'].isin(top_roles)].dropna(subset=['RemoteWork', 'DevType'])

    plt.figure(figsize=(12, 6))
    
    # Normalized 100% stacked bar plot
    crosstab_plot = pd.crosstab(role_df['DevType'], role_df['RemoteWork'], normalize='index') * 100
    crosstab_plot.plot(
        kind='barh', 
        stacked=True, 
        colormap='viridis', 
        figsize=(12, 6)
    )

    plt.title("Remote Work Breakdown across Top Developer Roles (%)", fontsize=13, fontweight="bold", pad=15)
    plt.xlabel("Percentage of Role (%)", fontsize=10)
    plt.ylabel("Developer Role (DevType)", fontsize=10)
    plt.legend(title="Remote Work Status", bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
```

`Employment` and `DevType` were exploded along semicolon delimiter into distinct observations. To preserve analytical integrity and avoid multi-counting skew, cross-tabulations were normalized row-wise (`normalize='index'`). This measures remote work distributions strictly within each individual employment cohort.

```python
# Split semicolon-separated values into individual rows
emp_exploded = (
    df[["RemoteWork", "Employment"]]
    .dropna(subset=["RemoteWork", "Employment"])
    .assign(Employment=df["Employment"].str.split(";"))
    .explode("Employment")
    .reset_index(drop=True)
)

# Strip any whitespace around separated tags
emp_exploded["Employment"] = emp_exploded["Employment"].str.strip()

# Cross-tabulate (Normalized by row to show percentage per employment type)
crosstab_emp = (
    pd.crosstab(
        emp_exploded["Employment"],
        emp_exploded["RemoteWork"],
        normalize="index",
    )
    * 100
)

display(crosstab_emp.round(2))
```

```python
# Explode DevType and reset index
dev_exploded = (
    df[["RemoteWork", "DevType"]]
    .dropna(subset=["RemoteWork", "DevType"])
    .assign(DevType=df["DevType"].str.split(";"))
    .explode("DevType")
    .reset_index(drop=True) 
)

dev_exploded["DevType"] = dev_exploded["DevType"].str.strip()

crosstab_roles = (
    pd.crosstab(
        dev_exploded["DevType"],
        dev_exploded["RemoteWork"],
        normalize="index",
    )
    * 100
)

display(crosstab_roles.round(2))
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Set clean aesthetic styling
sns.set_theme(style="whitegrid", font="sans-serif")
custom_colors = ["#4C72B0", "#55A868", "#C44E52"]  # Hybrid, Remote, In-person

# ---------------------------------------------------------
# Plot 1: Remote Work Preferences by Employment Type
# ---------------------------------------------------------
ax1 = crosstab_emp.plot(
    kind="barh",
    stacked=True,
    figsize=(10, 5),
    color=custom_colors,
    width=0.6,
)

plt.title(
    "Remote Work Setup by Employment Type (%)",
    fontsize=12,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Percentage of Cohort (%)", fontsize=10)
plt.ylabel("Employment Status", fontsize=10)
plt.xlim(0, 100)
plt.legend(
    title="Remote Work Status",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    frameon=True,
)

# Annotate percentages inside segments
for p in ax1.patches:
    width = p.get_width()
    if width > 5:  # Only label segments wider than 5%
        x = p.get_x() + width / 2
        y = p.get_y() + p.get_height() / 2
        ax1.annotate(
            f"{width:.1f}%",
            (x, y),
            ha="center",
            va="center",
            color="white",
            fontsize=9,
            fontweight="bold",
        )

plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# Plot 2: Remote Work Preferences by Top Job Roles (DevType)
# ---------------------------------------------------------
ax2 = crosstab_roles.sort_values(
    by="Remote", ascending=True
).plot(  # Sort by highest % remote
    kind="barh",
    stacked=True,
    figsize=(12, 6),
    color=custom_colors,
    width=0.7,
)

plt.title(
    "Remote Work Setup across Top Developer Roles (%)",
    fontsize=12,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Percentage of Role Cohort (%)", fontsize=10)
plt.ylabel("Developer Role (DevType)", fontsize=10)
plt.xlim(0, 100)
plt.legend(
    title="Remote Work Status",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    frameon=True,
)

# Annotate percentages inside segments
for p in ax2.patches:
    width = p.get_width()
    if width > 6:  # Only label visible segments
        x = p.get_x() + width / 2
        y = p.get_y() + p.get_height() / 2
        ax2.annotate(
            f"{width:.1f}%",
            (x, y),
            ha="center",
            va="center",
            color="white",
            fontsize=8,
            fontweight="bold",
        )

plt.tight_layout()
plt.show()
```

> **Cross-Tabulation Plot Takeaways:**
> 1. **Contractor vs. Full-Time Flexibility:** Independent contractors and freelancers display the highest proportion of fully remote arrangements, whereas full-time corporate employees lean heavily toward hybrid setups.
> 2. **Role-Driven On-Site Demands:** Hardware, embedded, and executive leadership roles maintain noticeably higher in-person proportions compared to software-centric roles (e.g., DevOps, Cloud, Backend Engineers), which favor remote and hybrid models.

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
# Define target countries
top_countries = [
    "United States of America",
    "Germany",
    "United Kingdom of Great Britain and Northern Ireland",
    "India",
    "Bulgaria",
    "China",
    "Czech Republic",
]

# Filter for valid records & top countries
lang_df = df[df["Country"].isin(top_countries)].dropna(
    subset=["LanguageHaveWorkedWith", "Country"]
)

# Explode semicolon-separated languages and reset index
lang_exploded = (
    lang_df[["Country", "LanguageHaveWorkedWith"]]
    .assign(
        Language=lang_df["LanguageHaveWorkedWith"].str.split(";")
    )
    .explode("Language")
    .reset_index(drop=True)
)

# Clean extra whitespaces
lang_exploded["Language"] = lang_exploded["Language"].str.strip()

# Identify global top 10 languages across selected countries
top_languages = (
    lang_exploded["Language"].value_counts().head(10).index
)
lang_filtered = lang_exploded[
    lang_exploded["Language"].isin(top_languages)
]

# Cross-tabulate: Calculate usage % of each language per country
crosstab_lang = (
    pd.crosstab(
        lang_filtered["Country"],
        lang_filtered["Language"],
        normalize="index",
    )
    * 100
)

# Rename long country names for plot readability
country_clean_map = {
    "United States of America": "USA",
    "United Kingdom of Great Britain and Northern Ireland": "UK",
}
crosstab_lang = crosstab_lang.rename(index=country_clean_map)

display(crosstab_lang.round(1))
```

```python
# Set figure style
sns.set_theme(style="white", font="sans-serif")
plt.figure(figsize=(12, 6))

# Plot Heatmap
ax = sns.heatmap(
    crosstab_lang,
    annot=True,
    fmt=".1f",
    cmap="Blues",
    cbar_kws={"label": "% of Developers Using Language"},
    linewidths=0.5,
    square=False,
)

plt.title(
    "Top Programming Language Adoption Across Countries (%)",
    fontsize=13,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Programming Language", fontsize=10, labelpad=10)
plt.ylabel("Country", fontsize=10)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
import pandas as pd

# Clean and shorten EdLevel labels for clean visualization readability
ed_map = {
    "Bachelor’s degree (B.A., B.S., B.Eng., etc.)": "Bachelor's",
    "Master’s degree (M.A., M.S., M.Eng., MBA, etc.)": "Master's",
    "Professional degree (JD, MD, etc.)": "Professional",
    "Professional degree (JD, MD, Ph.D, ed.D, etc.)": "Professional",
    "Doctoral degree (Ph.D., Ed.D., etc.)": "Doctorate",
    "Some college/university study without earning a degree": "Some College",
    "Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)": "High School",
    "Associate degree (A.A., A.S., etc.)": "Associate",
    "Primary/elementary school": "Primary School",
    "Something else": "Other",
}

df["EdLevel_Clean"] = df["EdLevel"].replace(ed_map)

# Explode Employment column on ';' and reset index
emp_ed_exploded = (
    df[["EdLevel_Clean", "Employment"]]
    .dropna(subset=["EdLevel_Clean", "Employment"])
    .assign(Employment=df["Employment"].str.split(";"))
    .explode("Employment")
    .reset_index(drop=True)
)

# Strip extra whitespace
emp_ed_exploded["Employment"] = emp_ed_exploded["Employment"].str.strip()

# Order education levels hierarchically
ed_order = [
    "Primary School",
    "High School",
    "Some College",
    "Associate",
    "Bachelor's",
    "Master's",
    "Doctorate",
    "Professional",
]
existing_ed_order = [
    ed for ed in ed_order if ed in emp_ed_exploded["EdLevel_Clean"].unique()
]

# Cross-tabulate: Normalize row-wise (percentages within each education tier)
crosstab_ed_emp = (
    pd.crosstab(
        emp_ed_exploded["EdLevel_Clean"],
        emp_ed_exploded["Employment"],
        normalize="index",
    )
    * 100
)

# Reorder index by educational hierarchy
crosstab_ed_emp = crosstab_ed_emp.reindex(existing_ed_order)

display(crosstab_ed_emp.round(2))
```

```python
sns.set_theme(style="whitegrid", font="sans-serif")

ax = crosstab_ed_emp.plot(
    kind="barh",
    stacked=True,
    figsize=(12, 6),
    colormap="Spectral",
    width=0.7,
)

plt.title(
    "Employment Type Distribution Across Educational Attainment Tiers (%)",
    fontsize=13,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Percentage within Education Cohort (%)", fontsize=10)
plt.ylabel("Education Level (EdLevel)", fontsize=10)
plt.xlim(0, 100)
plt.legend(
    title="Employment Status",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    frameon=True,
)

# Annotate values larger than 7%
for p in ax.patches:
    width = p.get_width()
    if width > 7:
        x = p.get_x() + width / 2
        y = p.get_y() + p.get_height() / 2
        ax.annotate(
            f"{width:.1f}%",
            (x, y),
            ha="center",
            va="center",
            color="white",
            fontsize=8,
            fontweight="bold",
        )

plt.tight_layout()
plt.show()
```
