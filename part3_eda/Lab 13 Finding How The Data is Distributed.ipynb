### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
## Write your code here
df.isna().sum().sort_values()/len(df)
```

```python
for col in df.columns:
    missing_pct = (df[col].isnull().sum() / len(df)) * 100

    if missing_pct == 0:
        continue

    # Low Missingness (< 25%): Safe to Impute
    if missing_pct < 25:
        if pd.api.types.is_numeric_dtype(df[col]):
            fill_val = (
                df[col].median()
                if abs(df[col].skew()) > 1.0
                else df[col].mean()
            )
            df[col] = df[col].fillna(fill_val)
        else:
            df[col] = df[col].fillna(df[col].mode()[0])
        print(f"Imputed '{col}' ({missing_pct:.1f}% missing) using standard strategy.")

    #  Moderate/High Categorical Missingness (>= 25%): Keep as 'Missing' category
    elif not pd.api.types.is_numeric_dtype(df[col]):
        df[col] = df[col].fillna("Missing / Unanswered")
        print(
            f"Encoded '{col}' ({missing_pct:.1f}% missing) as 'Missing / Unanswered'."
        )

    # High Numerical Missingness (>= 25%): Preserved (Do not distort distribution)
    else:
        print(
            f"Skipped imputing numeric '{col}' ({missing_pct:.1f}% missing) to prevent skew."
        )
```

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
# Set theme and figure size
sns.set_theme(style="whitegrid", font="sans-serif")
plt.figure(figsize=(10, 5))

# Filter out missing values dynamically for this visualizer
jobsat_valid = df.dropna(subset=["JobSat"])

# Render KDE Plot
ax = sns.kdeplot(
    data=jobsat_valid,
    x="JobSat",
    fill=True,
    color="#1f77b4",
    alpha=0.4,
    linewidth=2.5,
    bw_adjust=0.7,  # Smoothness adjustment
)

# Overlay Median line
median_sat = jobsat_valid["JobSat"].median()
plt.axvline(
    median_sat,
    color="#d62728",
    linestyle="--",
    linewidth=2,
    label=f"Median JobSat ({median_sat:.0f})",
)

# Titles and Axis Labels
plt.title(
    "Kernel Density Estimate (KDE) of Developer Job Satisfaction",
    fontsize=13,
    fontweight="bold",
    pad=15,
)
plt.xlabel(
    "Job Satisfaction Rating (0 = Extremely Dissatisfied, 10 = Extremely Satisfied)",
    fontsize=10,
)
plt.ylabel("Density", fontsize=10)
plt.xlim(0, 10)
plt.legend(loc="upper left", frameon=True)

plt.tight_layout()
plt.show()
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
# Explode currently used languages (LanguageHaveWorkedWith)
worked_exploded = (
    df["LanguageHaveWorkedWith"]
    .dropna()
    .str.split(";")
    .explode()
    .str.strip()
    .value_counts(normalize=True) * 100
).reset_index()
worked_exploded.columns = ["Language", "Pct_WorkedWith"]

# Explode desired languages (LanguageWantToWorkWith)
want_exploded = (
    df["LanguageWantToWorkWith"]
    .dropna()
    .str.split(";")
    .explode()
    .str.strip()
    .value_counts(normalize=True) * 100
).reset_index()
want_exploded.columns = ["Language", "Pct_WantToWorkWith"]

#  Merge both metrics into a single comparison DataFrame
lang_comp = pd.merge(worked_exploded, want_exploded, on="Language", how="inner")

# Filter for Top 12 languages by current usage
top_12_langs = lang_comp.sort_values(by="Pct_WorkedWith", ascending=False).head(12)

display(top_12_langs.round(2))
```

```python
# Set theme
sns.set_theme(style="whitegrid", font="sans-serif")
plt.figure(figsize=(12, 7))

# Melt DataFrame for Seaborn grouped plotting
plot_data = top_12_langs.melt(
    id_vars="Language", 
    value_vars=["Pct_WorkedWith", "Pct_WantToWorkWith"],
    var_name="Category", 
    value_name="Percentage"
)

# Rename legend categories
plot_data["Category"] = plot_data["Category"].replace({
    "Pct_WorkedWith": "Currently Used (Have Worked With)",
    "Pct_WantToWorkWith": "Desired (Want to Work With)"
})

# Plot
ax = sns.barplot(
    data=plot_data,
    y="Language",
    x="Percentage",
    hue="Category",
    palette=["#2b5c8f", "#d95f02"],
    width=0.7
)

plt.title("Programming Languages: Current Usage vs. Future Developer Demand (%)", fontsize=13, fontweight="bold", pad=15)
plt.xlabel("Percentage of Respondents (%)", fontsize=10)
plt.ylabel("Programming Language", fontsize=10)
plt.legend(title="", loc="lower right", frameon=True)

# Annotate values on the bars
for p in ax.patches:
    width = p.get_width()
    if width > 3:
        ax.annotate(
            f"{width:.1f}%",
            (width, p.get_y() + p.get_height() / 2.0),
            ha="left",
            va="center",
            fontsize=8,
            xytext=(3, 0),
            textcoords="offset points"
        )

plt.xlim(0, plot_data["Percentage"].max() * 1.12)
plt.tight_layout()
plt.show()
```

> **Language Usage vs. Desired Growth Insights:**
> 
> 1. **Pronounced Decline in JavaScript Share:**
>    * While **JavaScript** holds the largest current usage share at **12.73%**, its desired share drops steeply to **8.61%**. This represents the most significant contraction in developer mindshare across the dataset, signaling a desire to move beyond traditional JS setups.
> 
> 2. **Python as the Ultra-Stable Standard:**
>    * **Python** demonstrates remarkable consistency, moving from **9.05%** of current work to **9.07%** of desired work. Unlike web stack tools that fluctuate, Python remains a rock-solid, highly stable core language across data science, AI, and backend engineering.
> 
> 3. **Broad Contraction Across Established Tooling:**
>    * Languages like **HTML/CSS** (11.05% $\rightarrow$ 7.50%), **Java** (5.38% $\rightarrow$ 3.86%), and **Bash/Shell** (6.02% $\rightarrow$ 4.98%) also show noticeable drops in desired share, as developers express interest in exploring more specialized or modern syntax.

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
from scipy import stats
# Ensure YearsCodePro is mapped to numeric
if "YearsCodePro_Numeric" not in df.columns:
    years_map = {"Less than 1 year": 0.5, "More than 50 years": 51.0}
    df["YearsCodePro_Numeric"] = (
        df["YearsCodePro"]
        .replace(years_map)
        .pipe(pd.to_numeric, errors="coerce")
    )

# Drop missing pairs specifically for correlation calculation
corr_data = df.dropna(subset=["YearsCodePro_Numeric", "JobSat"])

x = corr_data["YearsCodePro_Numeric"]
y = corr_data["JobSat"]

# 4. Calculate Pearson Correlation (Linear)
pearson_corr, pearson_p = stats.pearsonr(x, y)

# 5. Calculate Spearman Correlation (Monotonic Rank)
spearman_corr, spearman_p = stats.spearmanr(x, y)

# 6. Display Results
print("=== Correlation Analysis: Experience vs. Job Satisfaction ===")
print(f"Sample Size (n): {len(corr_data):,}")
print(f"Pearson Correlation (r):  {pearson_corr:.4f}  (p-value: {pearson_p:.2e})")
print(f"Spearman Correlation (ρ): {spearman_corr:.4f}  (p-value: {spearman_p:.2e})")
```

> **Statistical Analysis: Experience vs. Job Satisfaction**
>
> * **Sample Size:** $n = 28,356$ paired responses (complete case analysis).
> * **Pearson Correlation ($r$):** $0.1042$ ($p = 2.94 \times 10^{-69}$)
> * **Spearman Rank Correlation ($\rho$):** $0.1192$ ($p = 3.07 \times 10^{-90}$)
>
> **Core Takeaways:**
> 1. **Minimal Effect Size:** Professional coding experience exhibits a very weak positive correlation with job satisfaction. With $R^2 \approx 1.1\%$, tenure alone is a poor predictor of how satisfied a developer is in their current role.
> 2. **Statistical vs. Practical Significance:** The ultra-low $p$-values reflect the high statistical power of the large sample size ($n > 28,000$), rather than a strong underlying physical or organizational relationship.

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
## Write your code here
import pandas as pd

# Clean and simplify EdLevel labels
ed_map = {
    "Bachelor’s degree (B.A., B.S., B.Eng., etc.)": "Bachelor's",
    "Master’s degree (M.A., M.S., M.Eng., MBA, etc.)": "Master's",
    "Professional degree (JD, MD, etc.)": "Professional",
    "Professional degree (JD, MD, Ph.D, ed.D, etc.)": "Professional",
    "Doctoral degree (Ph.D., Ed.D., etc.)": "Doctorate",
    "Some college/university study without earning a degree": "Some College",
    "Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)": "High School",
    "Associate degree (A.A., A.S., etc.)": "Associate",
    "Primary/elementary school": "Primary School",
    "Something else": "Other",
}

if "EdLevel_Clean" not in df.columns and "EdLevel" in df.columns:
    df["EdLevel_Clean"] = df["EdLevel"].replace(ed_map)

#  Explode Employment multi-select tags
emp_ed_df = (
    df[["EdLevel_Clean", "Employment"]]
    .dropna(subset=["EdLevel_Clean", "Employment"])
    .assign(Employment=df["Employment"].str.split(";"))
    .explode("Employment")
    .reset_index(drop=True)
)

emp_ed_df["Employment"] = emp_ed_df["Employment"].str.strip()

# Order education levels hierarchically
ed_order = [
    "Primary School",
    "High School",
    "Some College",
    "Associate",
    "Bachelor's",
    "Master's",
    "Doctorate",
    "Professional",
]
existing_ed_order = [
    ed for ed in ed_order if ed in emp_ed_df["EdLevel_Clean"].unique()
]

# Cross-tabulation normalized row-wise (% within each education level)
crosstab_ed_emp = (
    pd.crosstab(
        emp_ed_df["EdLevel_Clean"],
        emp_ed_df["Employment"],
        normalize="index",
    )
    * 100
).reindex(existing_ed_order)

display(crosstab_ed_emp.round(2))
```

```python
sns.set_theme(style="whitegrid", font="sans-serif")

# Plot 100% Stacked Horizontal Bar Chart
ax = crosstab_ed_emp.plot(
    kind="barh",
    stacked=True,
    figsize=(12, 6),
    colormap="Spectral",
    width=0.7,
)

plt.title(
    "Employment Status Distribution by Education Level (%)",
    fontsize=13,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Percentage within Educational Cohort (%)", fontsize=10)
plt.ylabel("Education Level (EdLevel)", fontsize=10)
plt.xlim(0, 100)
plt.legend(
    title="Employment Status",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    frameon=True,
)

# Annotate exact percentages for visible segments (> 7%)
for p in ax.patches:
    width = p.get_width()
    if width > 7:
        x = p.get_x() + width / 2
        y = p.get_y() + p.get_height() / 2
        ax.annotate(
            f"{width:.1f}%",
            (x, y),
            ha="center",
            va="center",
            color="white",
            fontsize=8,
            fontweight="bold",
        )

plt.tight_layout()
plt.show()
```

> **Education Level vs. Employment Status Insights:**
> 1. **Full-Time Employment Concentration:** Higher formal education degrees (Master's, Doctorate, Professional) show the highest concentration of full-time organizational employment.
> 2. **Self-Employment & Freelancing:** Lower formal degree tiers (High School, Some College) exhibit proportionally higher rates of independent contracting, freelancing, and part-time/self-employed arrangements.
