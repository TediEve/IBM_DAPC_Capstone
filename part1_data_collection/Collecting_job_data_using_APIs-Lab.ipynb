#### Write a function to get the number of jobs for the Python technology.
> **Note:** 
While completing this exercise, first retrieve the full dataset from the API endpoint.  
After retrieving the data, apply the required filtering and calculations locally in Python based on the given parameters.
  
 ##### The keys in the json are 
 * Job Title
 
 * Job Experience Required
 
 * Key Skills
 
 * Role Category
 
 * Location
 
 * Functional Area
 
 * Industry
 
 * Role 
 
You can also view  the json file contents  from the following URL.

https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json



```python
def get_number_of_jobs_T(technology):
    # Primary local endpoint & backup S3 URL
    local_url = "http://127.0.0.1:5000/data/all"
    backup_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"
    try:
        print("Attempting connection to local Flask app...")
        response = requests.get(local_url, timeout=10)
        response.raise_for_status()
        data = response.json()
        print("Successfully retrieved data from local Flask app!")
    except (requests.exceptions.RequestException, requests.exceptions.Timeout) as e:
        print(f"Local Flask server failed or timed out (Error: {e}).")
        print("Switching to backup S3 URL...")
        response = requests.get(backup_url)
        data = response.json()
    number_of_jobs = 0
    for job in data:
        skills = job.get("Key Skills", "")
        # Check if technology exists in the skills string (case-insensitive)
        if technology.lower() in str(skills).lower():
            number_of_jobs += 1
    return technology, number_of_jobs
```

#### Write a function to find number of jobs in US for a location of your choice


```python
 def get_number_of_jobs_L(location):
    # Primary local endpoint & backup S3 URL
    local_url = "http://127.0.0.1:5000/data/all"
    backup_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"
    try:
        print("Attempting connection to local Flask app...")
        response = requests.get(local_url, timeout=10)
        response.raise_for_status()
        data = response.json()
        print("Successfully retrieved data from local Flask app!")
    except (requests.exceptions.RequestException, requests.exceptions.Timeout) as e:
        print(f"Local Flask server failed or timed out (Error: {e}).")
        print("Switching to backup S3 URL...")
        response = requests.get(backup_url)
        data = response.json()
    number_of_jobs = 0
    for job in data:
        locations = job.get("Location", "")
        # Check if technology exists in the skills string (case-insensitive)
        if location.lower() in str(locations).lower():
            number_of_jobs += 1
    return location, number_of_jobs
```
